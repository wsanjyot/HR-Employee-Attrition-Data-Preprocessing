```python
# 1. Continuous numerical columns → Scale

# These represent quantities where numerical distance makes sense.

# Examples:

# Age
# DailyRate
# DistanceFromHome
# HourlyRate
# MonthlyIncome
# MonthlyRate
# TotalWorkingYears
# YearsAtCompany
# YearsInCurrentRole
# YearsSinceLastPromotion
# YearsWithCurrManager

# For example:

# Age: 25 → 26
# Age: 25 → 35

# The numerical difference has meaning.

# These are candidates for:

# StandardScaler()
# 2. Categorical columns → Encode

# These represent categories rather than quantities.

# For example:

# BusinessTravel
# Department
# EducationField
# Gender
# JobRole
# MaritalStatus
# OverTime

# Consider:

# Gender:
# Male
# Female

# We cannot give:

# Male = 1
# Female = 2

# and pretend that 2 > 1 has meaningful mathematical significance.

# So we use encoding.

# For nominal categorical variables, we'll generally use:

# OneHotEncoder()

# For example:

# Department

# Sales
# HR
# Research & Development

# becomes something conceptually like:

# Department_HR
# Department_Research & Development
# Department_Sales
# 3. Integer columns that LOOK numerical but aren't necessarily continuous

# This is the interesting part.

# Look at:

# Education: 5 unique values
# JobLevel: 5 unique values
# JobSatisfaction: 4 unique values
# PerformanceRating: 2 unique values

# They're int64, but they represent levels/ratings, not normal continuous measurements.

# For example:

# JobSatisfaction

# 1 = Low
# 2 = Medium
# 3 = High
# 4 = Very High

# The numbers represent ordered categories.

# This is called ordinal data.

# So we shouldn't automatically say:

# if dtype == "int64":
#     StandardScaler()

# because that would treat every integer column as continuous.

# Now look at these
# EmployeeCount: 1 unique value
# Over18: 1 unique value
# StandardHours: 1 unique value

# They have zero variation.

# For example:

# EmployeeCount

# 1
# 1
# 1
# 1
# 1
# ...

# There is nothing useful for a model to learn from this column.

# So these are candidates for removal.

# And this one is special
# EmployeeNumber: 1470 unique values

# You have 1470 employees and 1470 employee numbers.

# That means:

# EmployeeNumber
# 1001
# 1002
# 1003
# ...

# is essentially an identifier, not a meaningful feature.

# We shouldn't scale it just because it's an integer.

# So we'd normally remove it too.

# This is the main lesson

# Your dataset is teaching you something very important:

#                     Data Type
#                        ↓
#                  Is it enough?
#                        ↓
#                       NO
#                        ↓
#              Understand the column
#                        ↓
#        ┌───────────────┼────────────────┐
#        ↓               ↓                ↓
#   Continuous       Categorical       Identifier/
#    numerical        / Ordinal         Constant
#        ↓               ↓                ↓
#     Scaling          Encoding         Remove

# This is much more professional than simply doing:

# numerical_cols = df.select_dtypes(include="number").columns
# categorical_cols = df.select_dtypes(include="object").columns

# That method is useful as a starting point, but it can make mistakes when integer columns represent categories, ratings, IDs, or constants.
```

## column classifiatoin

Before column treansformation


columns to remove:

* EmployeeCount

* Over18

* StandardHours
* EmployeeNumber

```python
drop_cols = ["EmployeeCount", "EmployeeNumber" , "Over18" , "StandardHours"]
numerical_columns = [ "Age",
    "DailyRate",
    "DistanceFromHome",
    "HourlyRate",
    "MonthlyIncome",
    "MonthlyRate",
    "NumCompaniesWorked",
    "PercentSalaryHike",
    "TotalWorkingYears",
    "TrainingTimesLastYear",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsSinceLastPromotion",
    "YearsWithCurrManager"]

# added later

ordinal_cols = [
    "Education",
    "EnvironmentSatisfaction",
    "JobInvolvement",
    "JobLevel",
    "JobSatisfaction",
    "PerformanceRating",
    "RelationshipSatisfaction",
    "StockOptionLevel",
    "WorkLifeBalance"
]




categorical_columns = ["BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime"]
x_train = x_train.drop(columns = drop_cols)
x_test = x_test.drop(columns = drop_cols)
```

we did not include the ordinal columns, and ColumnTransformer by default drops columns that aren't specified in a transformer.

```python
# Education
# EnvironmentSatisfaction
# JobInvolvement
# JobLevel
# JobSatisfaction
# PerformanceRating
# RelationshipSatisfaction
# StockOptionLevel
# WorkLifeBalance


# A column can be numerical in storage (int64) but ordinal in meaning.
```

These columns are ordinal + numerical + discrete, so we should not treat them like ordinary continuous measurements, and we should not one-hot encode them by default.

we will decide later what to do whit this

## column transformation

```python
  #  X_train
  #                   │
  #                   ▼
  #         ┌───────────────────┐
  #         │ ColumnTransformer │
  #         └───────────────────┘
  #            │             │
  #            ▼             ▼
  #      Numerical        Categorical
  #         │                  │
  #         ▼                  ▼
  # StandardScaler      OneHotEncoder
  #         │                  │
  #         └────────┬─────────┘
  #                  ▼
  #         Processed X_train
```

```python
# Why handle_unknown="ignore"?

# Suppose the training data contains:

# JobRole
# Sales Executive
# Research Scientist
# Manager

# but the test data happens to contain a category that wasn't present in training.

# Without handle_unknown="ignore", the encoder can produce an error.

# With:

# handle_unknown="ignore"

# it safely handles unseen categories.

# This is a common choice in production preprocessing pipelines.
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
preprocessor = ColumnTransformer(transformers=[
    ("num" , StandardScaler() , numerical_columns),
    ("ord", StandardScaler(), ordinal_cols),  #added later
    ("cat" , OneHotEncoder(handle_unknown="ignore") , categorical_columns)
])

x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed = preprocessor.transform(x_test)
```
